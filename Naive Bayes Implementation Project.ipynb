```python
# ================================================
# 1) Import Required libraries
# ================================================

import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ================================================
# 2) Load Dataset
# ================================================

data = load_breast_cancer()

# Feature and Target

x = data.data
y = data.target

# Convert ro DataFrame for data feeling and better understanding

df = pd.DataFrame(x, columns=data.feature_names)
df['target'] = y

print("Dataset Shape", df.shape)
print("\nFirst 5 rows of dataset:\n")
print(df.head())
```

```python
# ================================================
# 3) Data information:
# ================================================

print("\nTarget Classes:")
print(data.target_names)

print("\nClass Distribution:")
print(df['target'].value_counts())

# ================================================
# 4) Split Dataset into Training and testing sets
# ================================================

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42, stratify=y
)

print("\nTraining Samples:", x_train.shape[0])

print("\nTesting Samples:", x_test.shape[0])
```

```python
# ================================================
# 5) Create and Train Naive Bayes Model
# ================================================
nb_model = GaussianNB()

# Train the model

nb_model.fit(x_train, y_train)

print("\nNaive Bayes model training completed.")
```

```python
# ================================================
# 6) make pcreditions
# ================================================
y_pred = nb_model.predict(x_test)

print("\nFirst 10 prediction:")
print(y_pred[:10])
```

```python
# ================================================
# 7) Model Evaluation.
# ================================================

# Accuracy 
accuracy = accuracy_score(y_test, y_pred)
print("\nModel Accuracy:", round(accuracy * 100, 2), "%")

# Confusion Matrix:

cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print (cm)
# Detailed Classification metric

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# ================================================
# 8) Interpretation of Confusion Matrix:
# ================================================

print("\nConfusion MAtrix Interpretation:")
print("Ture Negative (TN):", cm[0][0])
print("False Positive (FP):", cm[0][1])
print("Flase Negitive (FN):", cm[1][0])
print("True Positive (TP):", cm[1][1])
```

```python
# ================================================
# 9) Predict on New Sample
# ================================================

sample = x_test[0].reshape(1, -1)
prediction = nb_model.predict(sample)

print("\nPrediction for a new sample:",
     data.target_names[prediction][0])
```
